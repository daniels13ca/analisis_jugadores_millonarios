# Descarga de datos

Este notebook usa el paquete `millos_data` para buscar equipos y descargar la temporada completa sin duplicar logica.

```python
from pathlib import Path
import sys

def find_repo_root(start: Path) -> Path:
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / 'src' / 'millos_data').exists():
            return candidate
    return start.resolve()

ROOT = find_repo_root(Path.cwd())
SRC = ROOT / 'src'
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))
ROOT
```

```python
from millos_data import ApiConfig, download_season_matches, search_teams
import pandas as pd
```

## Configuracion

Define `FOOTBALL_API_KEY` en `.env` o en el entorno antes de ejecutar las celdas de descarga.
Si el notebook ya estaba abierto, reinicia el kernel para recargar el modulo actualizado.

```python
config = ApiConfig.from_env()
config
```

## Buscar equipos

```python
search_teams(config, 'Millonarios')
```

## Descargar una temporada completa

```python
SEASON = 2024
season_output = ROOT / f'Millonarios_{SEASON}_Stats_Detalladas'
summary = download_season_matches(config, season=SEASON, output_dir=season_output)
print(f"Equipo soportado: {summary['team_name']} ({summary['team_code']}, team_id={summary['team_id']})")
print(f"Partidos encontrados para descargar: {summary['fixtures_found']}")
print(f"Partidos ya descargados: {summary['already_downloaded']}")
print(f"Partidos pendientes antes de ejecutar: {summary['pending_before_download']}")
print(f"Partidos descargados en esta ejecucion: {summary['downloaded']}")
summary
```
